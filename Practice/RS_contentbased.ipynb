```python
import numpy as np 
import pandas as pd
```

```python
movies = pd.read_csv("movies.csv")
ratings = pd.read_csv("ratings_sample.csv")
```

```python
movies.head()
```

```python
ratings.head()
```

```python
movies["year"] = movies.title.str.extract(r"(\(\d\d\d\d\))", expand=False)
movies["year"] = movies.year.str.extract(r"(\d\d\d\d)", expand=False)
movies["title"] = movies.title.str.replace(r"(\(\d\d\d\d\))", "", regex=True)
movies["title"] = movies["title"].apply(lambda x: x.strip())
movies.head()
```

```python
movies["genres"] = movies.genres.str.split("|")
movies.head()
```

```python
moviesWithGenres_df = movies.copy()

for index, row in movies.iterrows():
    for genre in row["genres"]:
        moviesWithGenres_df.at[index, genre] = 1

moviesWithGenres_df = moviesWithGenres_df.fillna(0)
moviesWithGenres_df.head()
```

```python
ratings = ratings.drop("timestamp", axis=1)
ratings.head()
```

```python
userInput = [
    { "title":"Breakfast Club, The", "rating":5 },
    { "title":"Toy Story", "rating":3.5 },
    { "title":"Jumanji", "rating":2 },
    { "title":"Pulp Fiction", "rating":5 },
    { "title":"Akira", "rating":4.5}
]

inputMovies = pd.DataFrame(userInput)
inputMovies.head()
```

```python
inputId = movies[movies["title"].isin(inputMovies["title"].tolist())]
inputMovies = pd.merge(inputId, inputMovies, on="title")
inputMovies = inputMovies.drop("genres", axis=1)
inputMovies = inputMovies.drop("year", axis=1)
inputMovies.head()
```

```python
userMovies = moviesWithGenres_df[moviesWithGenres_df["movieId"].isin(inputMovies["movieId"].tolist())]
userMovies.head()
```
