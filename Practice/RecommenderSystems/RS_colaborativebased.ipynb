```python
import numpy as np
import pandas as pd
```

```python
movies_df = pd.read_csv("movies.csv")
rating_df = pd.read_csv("ratings_sample.csv")
```

```python
movies_df.head()
```

```python
movies_df["year"] = movies_df.title.str.extract(r"(\(\d\d\d\d\))", expand=False)
movies_df["year"] = movies_df.year.str.extract(r"(\d\d\d\d)", expand=False)
movies_df["title"] = movies_df.title.str.replace(r"(\(\d\d\d\d\))", "", regex=True)
movies_df["title"] = movies_df.title.apply(lambda x: x.strip())
movies_df.head()
```

```python
movies_df = movies_df.drop("genres", axis=1)
movies_df.head()
```

```python
rating_df.head()
```

```python
rating_df = rating_df.drop("timestamp", axis=1)
rating_df.head()
```

```python
userInput = [
    { "title":"Breakfast Club, The", "rating":5 },
    { "title":"Toy Story", "rating":3.5 },
    { "title":"Jumanji", "rating":2 },
    { "title":"Pulp Fiction", "rating":5 },
    { "title":"Akira", "rating":4.5}
]

inputMovies_df = pd.DataFrame(userInput)
inputMovies_df.head()
```

```python
inputId = movies_df[movies_df["title"].isin(inputMovies_df["title"].tolist())]
inputMovies_df = pd.merge(inputId, inputMovies_df)
inputMovies_df.head()
```

```python
inputMovies_df = inputMovies_df.drop("year", axis=1)
inputMovies_df.head()
```

```python
commonUsers_df = rating_df[rating_df["movieId"].isin(inputMovies_df["movieId"].tolist())]
commonUsers_df.head()
```

```python
commonUsersGroup = commonUsers_df.groupby("userId")
commonUsersGroup.get_group(13)
```

```python
commonUsersGroup = sorted(commonUsersGroup, key=lambda x: len(x[1]), reverse=True)
commonUsersGroup[:5]
```

```python
top100Users = commonUsersGroup[:100]
top100Users[0:5]
```

```python
personCorelationDict = {}

for name, group in top100Users:
    group = group.sort_values(by="movieId")
    inputMovies_df = inputMovies_df.sort_values(by="movieId")
    nRatings = len(group)
    similar_movies = inputMovies_df[inputMovies_df["movieId"].isin(group["movieId"].tolist())]
    similar_movies_ratings = similar_movies["rating"].tolist()
    all_ratings = group["rating"].tolist()
    Sxx = sum([i**2 for i in similar_movies_ratings]) - pow(sum(similar_movies_ratings), 2)/float(nRatings)
    Syy = sum([i**2 for i in all_ratings]) - pow(sum(all_ratings), 2)/float(nRatings)
    Sxy = sum( i*j for i, j in zip(similar_movies_ratings, all_ratings)) - sum(similar_movies_ratings)*sum(all_ratings)/float(nRatings)

    if Sxx != 0 and Syy != 0:
        personCorelationDict[name] = Sxy/np.sqrt(Sxx*Syy) 
    else:
        personCorelationDict[name] = 0
```

```python
personCorelationDict
```

```python
pearsonDF = pd.DataFrame.from_dict(personCorelationDict, orient="index")
pearsonDF.columns = ["similarityIndex"]
pearsonDF["userId"] = pearsonDF.index
pearsonDF.index = range(len(pearsonDF))
pearsonDF.head()
```

```python
topUsers = pearsonDF.sort_values(by="similarityIndex", ascending=False)[:50]
topUsers.head()
```

```python
topUsersRating = topUsers.merge(rating_df, left_on="userId", right_on="userId", how="inner")
topUsersRating.head()
```

```python
topUsersRating["weightedRating"] = topUsersRating["similarityIndex"]*topUsersRating["rating"]
topUsersRating.head()
```

```python
tempTopUsersRating = topUsersRating.groupby("movieId").sum()[["similarityIndex", "weightedRating"]]
tempTopUsersRating.columns = ["sum_similarityIndex", "sum_weightedRating"]
tempTopUsersRating.head()
```

```python
recommendation_df = pd.DataFrame()
recommendation_df["weighted average recommendation score"] = tempTopUsersRating["sum_weightedRating"]/tempTopUsersRating["sum_similarityIndex"]
recommendation_df["movieId"] = tempTopUsersRating.index
recommendation_df.head()
```

```python
recommendation_df = recommendation_df.sort_values(by="weighted average recommendation score", ascending=False)
recommendation_df.head(10)
```
