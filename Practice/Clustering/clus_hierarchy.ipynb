```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("cars_clus.csv")
print(df.size)
df.head()
```

```python
df[['sales', 'resale', 'type', 
    'price', 'engine_s', 'horsepow', 
    'wheelbas', 'width', 'length', 
    'curb_wgt', 'fuel_cap', 'mpg', 
    'lnsales', 'partition']] = df[['sales', 'resale', 'type', 
                                   'price', 'engine_s', 'horsepow', 
                                   'wheelbas', 'width', 'length', 
                                   'curb_wgt', 'fuel_cap', 'mpg', 
                                   'lnsales', 'partition']].apply(pd.to_numeric, errors='coerce')

df = df.dropna()
df = df.reset_index(drop=True)
df['kml'] = 236.25 / df['mpg']

print('shape of data set after cleaning: ', df.size)
df.head(5)
```

```python
feature_set = df[["engine_s", "horsepow", "wheelbas", "width", "length", "curb_wgt", "fuel_cap", "mpg", "kml"]]
```

```python
from sklearn.preprocessing import MinMaxScaler

X = feature_set.values
scaler = MinMaxScaler()
feature_mtx = scaler.fit_transform(X)
feature_mtx
```

```python
from sklearn.metrics.pairwise import euclidean_distances

distances = euclidean_distances(feature_mtx, feature_mtx)

distances
```

```python
from sklearn.cluster import AgglomerativeClustering

alggom = AgglomerativeClustering(6, linkage='complete')
alggom.fit(distances)

alggom.labels_
```

```python
df["cluster__"] = alggom.labels_
df.head()
```

```python
plt.Figure(figsize=(20, 12))

clusters = sorted(set(alggom.labels_))
colors = plt.cm.tab10(np.linspace(0, 1, max(len(clusters), 2)))

for color, cluster in zip(colors, clusters):
    subset = df[df.cluster__ == cluster]
    for i in subset.index:
        plt.text(subset.horsepow[i], subset.kml[i], str(subset["model"][i]), rotation=25)
        plt.scatter(subset.horsepow, subset.kml, s= subset.price*10, c=color)

plt.title('Clusters')
plt.xlabel('horsepow')
plt.ylabel('kml')
```
